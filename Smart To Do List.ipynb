```python
# Smart To-Do List AI
tasks = []

print("Namaste Sunitha! 📝")
print("Your To-Do List is ready!")
```

```python
# Kaam add karo
def add_task(task):
    tasks.append(task)
    print(f"✅ Added: {task}")

add_task("Learn Python")
add_task("Make chai")
add_task("Call Amma")

print("\nYour tasks:")
for i, t in enumerate(tasks, 1):
    print(f"{i}. {t}")
```

```python
# Kaam khatam karo
def done_task(task_number):
    done = tasks.pop(task_number - 1)
    print(f"🎉 Great Job Sunitha! Done: {done}")
    print(f"Bache hue kaam: {len(tasks)}")

# Chai bana li, ab usko khatam karo
done_task(2)

print("\nAb bache hue kaam:")
for i, t in enumerate(tasks, 1):
    print(f"{i}. {t}")
```

```python
# Sab kaam khatam karo
done_task(1)  # Learn Python done
done_task(1)  # Call Amma done

if len(tasks) == 0:
    print("\n🏆 WOW Sunitha! Sab kaam khatam!")
    print("Aaj tum Superstar ho! ⭐")
else:
    print(f"\nAbhi {len(tasks)} kaam bache hai")
```
